```python
processed_df = pd.read_csv(
    "../data/processed/places_county_clean.csv"
)

processed_df.head()
```

```python
processed_df["measure"].value_counts()
```

```python
measure_summary = (
    processed_df
    .groupby(["category", "measure"])
    .agg(
        observations=("data_value", "count"),
        mean_prevalence=("data_value", "mean"),
        median_prevalence=("data_value", "median"),
        min_prevalence=("data_value", "min"),
        max_prevalence=("data_value", "max")
    )
    .reset_index()
)

measure_summary
```

```python
measure_summary.sort_values(
    "mean_prevalence",
    ascending=False
).head(20)
```

```python
processed_df["datavaluetypeid"].value_counts()
```

```python
prevalence_comparison = (
    processed_df
    .groupby(
        ["measure", "datavaluetypeid"]
    )["data_value"]
    .mean()
    .reset_index()
)

prevalence_comparison
```

```python
prevalence_comparison.sort_values(
    "measure"
)
```

```python
state_summary = (
    processed_df
    .groupby(["stateabbr", "statedesc"])
    .agg(
        mean_prevalence=("data_value", "mean"),
        median_prevalence=("data_value", "median"),
        measures=("measureid", "nunique"),
        counties=("locationid", "nunique")
    )
    .reset_index()
)

state_summary.sort_values(
    "mean_prevalence",
    ascending=False
).head(10)
```

```python
age_adjustment = (
    processed_df
    .pivot_table(
        index="measure",
        columns="datavaluetypeid",
        values="data_value",
        aggfunc="mean"
    )
    .reset_index()
)

age_adjustment["difference"] = (
    age_adjustment["CrdPrv"]
    - age_adjustment["AgeAdjPrv"]
)

age_adjustment.sort_values(
    "difference",
    ascending=False
)
```

```python
age_adjustment.sort_values(
    "difference",
    ascending=False
).head(10)
```

```python
age_adjustment.sort_values(
    "difference",
    ascending=True
).head(10)
```

```python
county_counts = (
    processed_df
    .groupby(["stateabbr", "statedesc"])["locationid"]
    .nunique()
    .reset_index(name="county_count")
)

county_counts.sort_values(
    "county_count",
    ascending=False
).head(15)
```

```python
indicator_metadata = (
    processed_df[
        [
            "measureid",
            "measure",
            "category",
            "datavaluetypeid"
        ]
    ]
    .drop_duplicates()
    .sort_values(["category", "measureid"])
)

indicator_metadata
```

```python
indicator_metadata[
    indicator_metadata["datavaluetypeid"] == "AgeAdjPrv"
]
```

```python
measure_variability = (
    processed_df[
        processed_df["datavaluetypeid"] == "AgeAdjPrv"
    ]
    .groupby(
        ["measureid", "measure", "category"]
    )
    .agg(
        mean=("data_value", "mean"),
        median=("data_value", "median"),
        std=("data_value", "std"),
        minimum=("data_value", "min"),
        maximum=("data_value", "max")
    )
    .reset_index()
)

measure_variability["range"] = (
    measure_variability["maximum"]
    - measure_variability["minimum"]
)

measure_variability.sort_values(
    "range",
    ascending=False
).head(15)
```
